```python
import matplotlib.pyplot as plt

tokens = list(range(1_000_000))

precio = []

for t in tokens:

    if t <= 272_000:
        precio.append((8.16 / 272000) * t)

    else:
        precio.append((36.84 / 728000) * (t - 272000) + 8.16)


plt.figure(figsize=(10,6))

plt.plot(tokens, precio)

plt.xlabel("Cantidad de Tokens")
plt.ylabel("Precio ($)")
plt.title("Costo de Tokens IA")

plt.grid(True)

plt.show()
```

# 2. Encontrar todas las raíces de la ecuación $f(x)=1/4(x^3+3x^2−6x−8)$ usando el método de la bisección

```python
import numpy as np
import matplotlib.pyplot as plt

# Función
def f(x):
    return (x**3 + 3*x**2 - 6*x - 8)/4


# Método de bisección
def biseccion(a, b, iteraciones):

    if f(a) * f(b) > 0:
        print(f"No hay cambio de signo en [{a}, {b}]")
        return

    print(f"\nIntervalo inicial: [{a}, {b}]")

    for i in range(iteraciones):

        c = (a + b) / 2

        print(f"Iteración {i+1}")
        print(f"a = {a:.6f}")
        print(f"b = {b:.6f}")
        print(f"c = {c:.6f}")
        print(f"f(c) = {f(c):.6f}")

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print(f"\nRaíz aproximada: {c:.6f}")


# Intervalos dados
rangos = [
    (-5, 3),
    (-4.7, 2.5),
    (-4.5, 2.7),
    (-3, -2)
]

# Ejecutar bisección
for a, b in rangos:
    biseccion(a, b, 5)


# Gráfica
x = np.linspace(-6, 4, 1000)
y = f(x)

plt.figure(figsize=(10,6))
plt.plot(x, y, label='f(x)')
plt.axhline(0)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Método de Bisección")
plt.grid(True)
plt.legend()

plt.show()
```
