```python
import matplotlib.pyplot as plt

tokens = list(range(1_000_000))

precio = []

for t in tokens:
    if t <= 272_000:
        costo = (t / 1_000_000) * 15
    else:
        costo = ((272_000 / 1_000_000) * 15) + \
                 (((t - 272_000) / 1_000_000) * 10)

    precio.append(costo)

print(precio[:10])
print(precio[-5:])

plt.plot(tokens, precio)
plt.xlabel("Tokens")
plt.ylabel("Precio ($)")
plt.title("Costo de tokens IA")
plt.grid(True)

plt.show()
```
